# 01 Data Ingestion
Load the raw Bluestock CSV sources and record row/column quality checks.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
RAW = ROOT / 'data' / 'raw'
files = sorted(RAW.glob('*.csv'))
inventory = []
for path in files:
    frame = pd.read_csv(path)
    inventory.append({'file': path.name, 'rows': len(frame), 'columns': len(frame.columns), 'missing_cells': int(frame.isna().sum().sum())})
inventory = pd.DataFrame(inventory).sort_values('file')
inventory
```
