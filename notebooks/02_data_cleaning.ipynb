# 02 Data Cleaning
Standardize date fields, remove duplicate rows, and validate key identifiers before analytics.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
RAW = ROOT / 'data' / 'raw'
PROCESSED = ROOT / 'data' / 'processed'
PROCESSED.mkdir(parents=True, exist_ok=True)
date_columns = {'date', 'month', 'transaction_date', 'launch_date'}
for source in sorted(RAW.glob('*.csv')):
    frame = pd.read_csv(source)
    frame.columns = [column.strip().lower() for column in frame.columns]
    for column in date_columns.intersection(frame.columns):
        frame[column] = pd.to_datetime(frame[column], errors='coerce')
    frame = frame.drop_duplicates()
    target = PROCESSED / f'cleaned_{source.name}'
    frame.to_csv(target, index=False, date_format='%Y-%m-%d')
print('Cleaned files:', len(list(PROCESSED.glob('cleaned_*.csv'))))
```
