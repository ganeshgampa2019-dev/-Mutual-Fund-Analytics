# 03 EDA Analysis
Explore industry scale, fund-house concentration, investor behavior, and category flows.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd(); DATA = ROOT / 'data' / 'processed'
aum = pd.read_csv(DATA / 'cleaned_03_aum_by_fund_house.csv', parse_dates=['date'])
transactions = pd.read_csv(DATA / 'cleaned_08_investor_transactions.csv', parse_dates=['transaction_date'])
category = pd.read_csv(DATA / 'cleaned_05_category_inflows.csv', parse_dates=['month'])
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
aum.groupby('date').aum_lakh_crore.sum().plot(ax=axes[0], title='Industry AUM trend')
aum.groupby('fund_house').aum_lakh_crore.max().nlargest(10).sort_values().plot.barh(ax=axes[1], title='Top fund houses by AUM')
transactions.groupby('transaction_type').amount_inr.sum().plot.pie(ax=axes[2], autopct='%1.0f%%', title='Transaction value mix')
plt.tight_layout()
```
