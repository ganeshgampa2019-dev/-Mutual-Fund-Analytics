# 04 Performance Analytics
Calculate annualized return, volatility, Sharpe, Beta, maximum drawdown, and historical VaR from daily NAV observations.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
metrics = pd.read_csv(ROOT / 'data' / 'fund_metrics.csv')
metrics.sort_values('sharpe_ratio', ascending=False).head(10)
```

## Formula notes
- Annualized return uses 252 trading days, not calendar days.
- Sharpe = mean daily excess return / daily standard deviation * sqrt(252), with a 6% annual risk-free rate.
- Beta = covariance(fund, benchmark) / variance(benchmark).
- Historical daily VaR at 95% is the 5th percentile of daily returns; it is reported as a negative loss threshold.
